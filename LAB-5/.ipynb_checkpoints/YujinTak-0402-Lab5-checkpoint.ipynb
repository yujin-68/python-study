## LAB5-5: 두 수를 전달받아 연산을 수행하는 함수

1. 두 개의 매개변수 a, b를 받아서 두 수의 차를 구하여 출력하는 `print_sub(a, b)` 함수를 구현하여라. `print_sub(10, 20)`을 호출한 결과 다음과 같은 출력이 나타나도록 하여라.
```
10과 20의 차는 -10입니다.
```

2. 두 개의 매개변수 a, b를 받아서 두 수의 곱을 구하여 출력하는 `print_mult(a, b)` 함수를 구현하여라. `print_mult(10, 20)`을 호출한 결과 다음과 같은 출력이 나타나도록 하여라.

## LAB5-6: 함수의 사용

1. 다음 2차 방정식의 근을 [코드 5-10]의 함수를 사용하여 출력하여라.
   - (1) x² + 4x − 21 = 0
   - (2) x² − 6x + 8 = 0

2. 삼각형의 면적을 구하기 위하여 밑변과 높이를 사용하려고 한다. 두 개의 매개변수 width, height를 받아서 삼각형의 면적을 출력하는 함수 `print_area(width, height)`를 구현하여라. 이때 `print_area(10, 20)`을 호출하여 다음과 같은 출력이 나타나도록 하여라.
```
밑변 10, 높이 20인 삼각형의 면적은 : 100
```

````python
import math

def find_root(a, b, c):
    discriminant = b**2 - 4*a*c
    root1 = (-b + math.sqrt(discriminant)) / (2*a)
    root2 = (-b - math.sqrt(discriminant)) / (2*a)
    return root1, root2

# (1) x² + 4x - 21 = 0
r1, r2 = find_root(1, 4, -21)
print(f'x² + 4x - 21 = 0 의 근: {r1}, {r2}')

# (2) x² - 6x + 8 = 0
r1, r2 = find_root(1, -6, 8)
print(f'x² - 6x + 8 = 0 의 근: {r1}, {r2}')

def print_area(width, height):
    area = width * height / 2
    print(f'밑변 {width}, 높이 {height}인 삼각형의 면적은 : {int(area)}')

print_area(10, 20)  # 삼각형의 면적을 구하는 함수 호출
````

## LAB5-7: 원의 면적과 둘레를 반환하는 함수

1. 원의 면적과 둘레를 구하기 위하여 원의 반지름을 사용하려고 한다. 이 때 사용할 `circle_area_circum(radius)` 함수는 다음과 같이 반지름(radius) 10을 입력으로 받아서 면적(area)과 둘레(circum)의 두 값을 반환하도록 구현하여라.

```python
radius = 10
area, circum = circle_area_circum(radius)  # 원의 면적과 둘레를 반환
```

실행결과:
```
반지름 10인 원의 면적은 314.0, 원의 둘레는 62.8
```

````python
def circle_area_circum(radius):
    pi = 3.14
    area = pi * radius ** 2
    circum = 2 * pi * radius
    return area, circum

radius = 10
area, circum = circle_area_circum(radius)  # 원의 면적과 둘레를 반환
print(f'반지름 {radius}인 원의 면적은 {area}, 원의 둘레는 {circum}')
````

## LAB5-8: 다수의 결과를 반환하는 함수 만들기

1. 어떤 정수 n과 m을 입력하면, n의 배수 m개를 반환하는 `multiples(n, m)` 함수를 구현하고 다음과 같이 호출하여 결과를 출력하여라.

```python
r1, r2, r3, r4 = multiples(3, 4)       # 3의 배수 4개를 구하라.
print(r1, r2, r3, r4)
r1, r2, r3, r4, r5 = multiples(2, 5)  # 2의 배수 5개를 구하라.
print(r1, r2, r3, r4, r5)
```

실행결과:
```
3 6 9 12
2 4 6 8 10
```

````python
def multiples(n, m):
    result = []
    for i in range(1, m + 1):
        result.append(n * i)
    return result

r1, r2, r3, r4 = multiples(3, 4)       # 3의 배수 4개를 구하라.
print(r1, r2, r3, r4)
r1, r2, r3, r4, r5 = multiples(2, 5)  # 2의 배수 5개를 구하라.
print(r1, r2, r3, r4, r5)
````

```python
def print_name(honorifics, first_name, last_name):
    ''' 키워드 인자를 이용한 출력용 프로그램 '''
```

## LAB5-10: 가변 인자의 활용

1. 가변 인자를 사용하는 `sum_nums()` 함수를 수정하여 인자들을 튜플 형식으로 출력한 후 모든 값들의 합과 평균을 출력하시오.

2. 가변 인자를 사용하는 `min_nums()` 함수를 구현하시오. 이 함수는 수들을 가변 인자로 받아 최솟값을 반환한다. 이 함수의 호출문은 다음과 같은 경우
```python
min_nums(20, 40, 50, 10)
```
다음과 같은 결과로 출력되도록 하시오.
```
최솟값 = 10
```

````python
# 1. sum_nums 함수
def sum_nums(*args):
    print('인자들:', args)
    total = sum(args)
    avg = total / len(args)
    print(f'합계 = {total}, 평균 = {avg}')

sum_nums(10, 20, 30)
sum_nums(10, 20, 30, 40, 50)

# 2. min_nums 함수
def min_nums(*args):
    return min(args)

print('최솟값 =', min_nums(20, 40, 50, 10))
````

## LAB5-11: 다양한 문자열 포매팅

1. 위의 코드를 수정하여 이름, 나이, 직업, 사는 곳까지 다음과 같이 화면에 출력하려고 한다.

a) 출력문에서 `format()` 메소드를 사용하여 출력하여라.

b) 출력문에서 f-문자열을 사용하여 출력하여라.

c) 출력문에서 `format()` 메소드나 f-문자열을 사용하지 않고 쉼표로 구분하여 출력하여라.

````python
name = '홍길동'
age = 20
job = '학생'
city = '서울'

# a) format() 메소드 사용
print('이름: {}, 나이: {}, 직업: {}, 사는 곳: {}'.format(name, age, job, city))

# b) f-문자열 사용
print(f'이름: {name}, 나이: {age}, 직업: {job}, 사는 곳: {city}')

# c) 쉼표로 구분
print('이름:', name, '나이:', age, '직업:', job, '사는 곳:', city)
````
