## 3번. list1, list2가 있을 경우 이중 for문을 이용하여 각 원소을 곱하기

```python
lst1 = [3,5,7]
lst2 = [2,3,4,5,6]

for i in lst1 :
    for j in lst2 :
        result = i * j
        print(f'{i} * {j} = {result}')
```

## 5번. list1과 list2를 글자 조합하기

```python
lst1 = ['I like','I love']
lst2 = ['pan', 'kiwi', 'es']

for i in lst1:
    for j in lst2 :
        print(i + j)
```

## 9번. max() 사용하지 않고 리스트 원소들 중 최댓값 찾기

```python
lst = [10, 20, 30, 50, 60]
result = 0
for i in range(len(lst)):
    if lst[i] < lst[i-1]:
        result = lst[i-1]
print(result)
```

## 13번. 표준편차 구하기 

```python
list1 = list(map(int, input("10개의 수를 입력하세요 :").split()))
total = sum(list1)
mean = total / len(list1)
a = 0
for i in list1:
    a += (i - mean) ** 2
st_divi = (a / 10) **(1/2)
print("합 :",total)
print("평균 :",mean)
print("표준편차 : {:0.2f}".format(st_divi))
```
