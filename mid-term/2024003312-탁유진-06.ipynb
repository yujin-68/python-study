1. 내장함수를 사용하지 않는 방식

```python
num_list = [-6, 3, 7, 9, -3, 5, -7, 1]

def find_max(nlist):
    max_num = nlist[0]
    for num in nlist:
        if num > max_num:
            max_num = num
    return max_num

def find_min(nlist):
    min_num = nlist[0]
    for num in nlist:
        if num < min_num:
            min_num = num
    return min_num

def cal(nlist):
    return find_max(nlist) + find_min(nlist)

cal(num_list) #테스트용
```

2. 내장함수를 사용하는 방식

```python
num_list = [-6, 3, 7, 9, -3, 5, -7, 1]

def cal(nlist):
    max_num = max(nlist)
    min_num = min(nlist)
    return max_num + min_num

cal(num_list) #테스트용
```
