입력받은 자연수들의 산술 평균과 표준편차를 출력

```python
str_num_list = []
int_num_list = []

str_num_list = input("숫자를 공백으로 구분하여 입력해주세요.").split()
for str_num in str_num_list:
    int_num_list.append(int(str_num))

#산술평균을 구하는 코드
average = sum(int_num_list) / len(int_num_list)
#표준편차를 구하는 코드
result = []
for num in int_num_list:
    result.append(((num - average)**2)**0.5)

print("입력한 숫자의 산술평균은 {0}, 표준편차는 {1}입니다.".format(average, result))
```
