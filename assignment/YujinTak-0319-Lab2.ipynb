## Lab 2-3

```python
pi = 3.14159
radius = 8

circle_area = pi * radius ** 2
circle_perimeter = 2 * pi * radius

print('원의 반지름', radius)
print('원의 면적', circle_area)
print('원의 둘레', circle_perimeter)
```
