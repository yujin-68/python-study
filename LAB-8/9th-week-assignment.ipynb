### 문제 1. datetime으로 오늘 날짜와 현재 시간 출력하기
- `datetime` 모듈을 사용해 현재 날짜/시간을 출력하고, 시각은 오전/오후 12시간 형식으로 표시한다.

```python
from datetime import datetime

now = datetime.now()

hour_24 = now.hour
ampm = "오전" if hour_24 < 12 else "오후"
hour_12 = hour_24 % 12
if hour_12 == 0:
    hour_12 = 12

print(f"오늘의 날짜 : {now.year}년 {now.month}월 {now.day}일")
print(f"현재시간 : {ampm} {hour_12}시 {now.minute}분 {now.second}초")
```

### 문제 2. 남은 날짜 계산하기
- 오늘로부터 특정 날짜(2027년 크리스마스, 2036년 1월 1일, 다음 생일)까지 남은 일/시간을 계산한다.

```python
from datetime import datetime

now = datetime.now()

def print_remaining_time(target, label):
    delta = target - now
    total_seconds = int(delta.total_seconds())
    days = total_seconds // (24 * 3600)
    hours = (total_seconds % (24 * 3600)) // 3600
    print(f"{label}까지는 {days}일 {hours}시간 남았습니다.")

print(f"오늘은 {now.year}년 {now.month}월 {now.day}일입니다.")

# (1) 2027년 크리스마스까지
christmas_2027 = datetime(2027, 12, 25)
print_remaining_time(christmas_2027, "2027년 크리스마스")

# (2) 2036년 1월 1일까지
newyear_2036 = datetime(2036, 1, 1)
print_remaining_time(newyear_2036, "2036년 새해")

# (3) 다가오는 내 생일까지 (예시: 8월 29일)
birth_month, birth_day = 8, 29
next_birthday = datetime(now.year, birth_month, birth_day)
if next_birthday <= now:
    next_birthday = datetime(now.year + 1, birth_month, birth_day)
print_remaining_time(next_birthday, f"{next_birthday.year}년 생일")
```

### 문제 3. timedelta로 날짜 계산하기
- 오늘로부터 1000일 후 날짜를 구하고, 처음 만난 날짜를 입력받아 100일 기념일을 계산한다.

```python
from datetime import datetime, timedelta

# (1) 오늘로부터 1000일 후 날짜
today = datetime.now().date()
after_1000_days = today + timedelta(days=1000)
print(f"오늘 날짜: {today}")
print(f"1000일 후 날짜: {after_1000_days}")

# (2) 커플 100일 기념일
# 입력 형식 예: 2019 3 30
year, month, day = map(int, input("처음으로 사귄 연도와 월, 일을 입력하세요 (YYYY M D): ").split())
start_day = datetime(year, month, day).date()
anniversary_100 = start_day + timedelta(days=99)  # 시작일을 1일째로 보면 +99

print(f"100일 기념일은 : {anniversary_100.year}년 {anniversary_100.month}월 {anniversary_100.day}일입니다.")
```

### 문제 4. math 모듈을 이용한 계산
- `math`와 반복문으로 거듭제곱, 라디안 변환, sin 값을 10도 간격으로 출력한다.

```python
import math

# (1) 4의 2승 ~ 10승
print("[1] 4의 거듭제곱")
for n in range(2, 11):
    print(f"4**{n} = {math.pow(4, n)}")

# (2) 0도 ~ 180도를 10도 단위로, 라디안 값 출력
print("\n[2] degree -> radian")
for degree in range(0, 181, 10):
    rad = math.radians(degree)
    print(f"{degree:3d} degree = {rad:.3f} radian")

# (3) sin 값 출력
print("\n[3] sin 값")
for degree in range(0, 181, 10):
    rad = math.radians(degree)
    print(f"sin({degree:3d}) = {math.sin(rad):.2f}")
```

### 문제 5. 랜덤 번호 생성기
- `randrange()`로 0~100의 5의 배수 3개를 뽑고, `sample()`로 1~10 사이 정수 3개를 뽑는다.

```python
import random

# (1) 0~100 이하의 정수 중 5의 배수 3개 (randrange 사용)
multiples = []
while len(multiples) < 3:
    num = random.randrange(0, 101, 5)
    if num not in multiples:
        multiples.append(num)

print(f"0에서 100 이하의 정수 중에서 5의 배수: {multiples}")

# (2) 1~10 사이 임의의 정수 3개 (sample 사용)
nums = random.sample(range(1, 11), 3)
print(f"1에서 10 사이의 임의의 정수: {nums}")
```

### 문제 6. 로또 번호 얻기
- 랜덤 로또 번호 5세트를 생성하고, 당첨 번호 6개와 보너스 번호 1개를 출력한다.

```python
import random

# (1) 매주 추천 로또 번호 5세트
print("이번 주의 추천 로또 번호")
for _ in range(5):
    lotto = sorted(random.sample(range(1, 46), 6))
    print(lotto)

# (2) 당첨 번호 + 보너스 번호
winning = sorted(random.sample(range(1, 46), 6))
bonus_pool = [n for n in range(1, 46) if n not in winning]
bonus = random.sample(bonus_pool, 1)

print("\n이번 주의 당첨 로또 번호 + 보너스")
print(f"{winning} + {bonus}")
```
