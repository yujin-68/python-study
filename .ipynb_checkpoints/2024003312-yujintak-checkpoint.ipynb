Quize 12
반복문으로 최고/최저/평균 직접 구하기

```python
#🔴반복문으로 최대값 구하기
def get_highest(scores):
    highest = scores[0] #첫 번째를 초기값으로
    for score in scores:
        if score > highest: highest = score
    return highest

def get_lowest(scores):
    lowest = scores[0]
    for score in scores:
        if score < lowest: lowest = score
    return lowest

def get_average(scores):
    total = 0
    n = len(scores)
    for score in scores: total += score
    average = total / n
    return average

def get_failing(scores, passing_score=60):
    failing_list = []
    for score in scores:
        if score < passing_score: failing_list.append(score)
    return failing_list

student_score_list = [10,20,40,15,60]
get_failing(student_score_list)
```

```python
#🔴리스트.insert(위치, 삽입할 요소)
def restock(inventory, item):
    inventory.insert(0,item)
    return inventory

#🔴리스트.pop() 가장 마지막 요소 삭제 후 반환
def ship(inventory):
    if len(inventory) == 0: 
        print("출고 실패. 재고 없음")
        return None
    item = inventory.pop()
    print(f"{item} 출고 완료.")
    return item

def check_stock(inventory, threshold=3):
    if len(inventory) < threshold: 
        print("재고 부족. {threshold}개 미만")
    return None

def get_oldest(inventory):
    return inventory[-1]

inv = []
restock(inv, "사과"); restock(inv, "바나나"); restock(inv, "딸기")
print(inv)
print(get_oldest(inv))
ship(inv)
```
