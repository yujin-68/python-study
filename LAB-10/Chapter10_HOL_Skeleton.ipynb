## 🧪 Lab 7. 캡슐화와 setter/getter

### 📋 과제
`BankAccount` 클래스를 만드세요.
- 속성: `owner`, `balance`
- 메소드:
  - `deposit(amount)`: 양수일 때만 입금, 아니면 `"잘못된 금액입니다."`
  - `withdraw(amount)`: 양수이고 잔액보다 작거나 같을 때만 출금
    - 음수/0이면 `"잘못된 금액입니다."`
    - 잔액 부족이면 `"잔액 부족입니다."`
  - `get_balance()`: 잔액 반환

### ✅ 예상 출력
```
입금 후 잔액: 15000
잘못된 금액입니다.
잔액: 15000
잔액 부족입니다.
잔액: 15000
출금 후 잔액: 8000
```

````python
class BankAccount:
    def __init__(self, owner, balance):
        # TODO 7-1: self.owner, self.balance 초기화
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        # TODO 7-2: amount > 0 이면 잔액 증가, 아니면 메시지 출력
        if amount > 0:
            self.balance += amount
        else:
            print("잘못된 금액입니다.")
    
    def withdraw(self, amount):
        # TODO 7-3: 조건 분기
        # - amount <= 0: "잘못된 금액입니다."
        # - amount > self.balance: "잔액 부족입니다."
        # - 그 외: self.balance -= amount
        if amount <= 0:
            print("잘못된 금액입니다.")
        elif amount > self.balance:
            print("잔액 부족입니다.")
        else:
            self.balance -= amount
    
    def get_balance(self):
        # TODO 7-4: 잔액 반환
        return self.balance


# 시나리오 테스트 (이미 작성됨 - 수정 X)
account = BankAccount("이서연", 10000)

account.deposit(5000)
print("입금 후 잔액:", account.get_balance())

account.deposit(-3000)
print("잔액:", account.get_balance())

account.withdraw(20000)
print("잔액:", account.get_balance())

account.withdraw(7000)
print("출금 후 잔액:", account.get_balance())
````

## 🧪 Lab 9. 특수 메소드 - 연산자 오버로딩

### 📋 과제
화폐를 표현하는 `Money` 클래스를 만드세요.
- `__init__(self, won)`
- `__str__`: `"₩1,000"` 형식 (천 단위 콤마)
- `__add__`: 두 Money를 더한 새 Money 반환
- `__sub__`: 결과가 음수면 `ValueError` 발생
- `__eq__`: `won` 값이 같으면 True

### ✅ 예상 출력
```
₩8,000
₩2,000
True
오류 발생: 결과가 음수가 될 수 없습니다.
```

### 💡 힌트
- 천 단위 콤마: `f"₩{self.won:,}"`
- 예외 발생: `raise ValueError("...")`

````python
class Money:
    def __init__(self, won):
        # TODO 9-1
        self.won = won
    
    def __str__(self):
        # TODO 9-2: "₩1,000" 형식 반환
        return f"₩{self.won:,}"
    
    def __add__(self, other):
        # TODO 9-3: Money(self.won + other.won) 반환
        return Money(self.won + other.won)
    
    def __sub__(self, other):
        # TODO 9-4: 음수면 ValueError, 아니면 Money 반환
        result = self.won - other.won
        if result < 0:
            raise ValueError("결과가 음수가 될 수 없습니다.")
        return Money(result)
    
    def __eq__(self, other):
        # TODO 9-5
        return isinstance(other, Money) and self.won == other.won


# 테스트
a = Money(5000)
b = Money(3000)

print(a + b)
print(a - b)
print(a == Money(5000))

try:
    print(b - a)
except ValueError as e:
    print("오류 발생:", e)
````

## 🧪 Lab 10. 상속 - 부모/자식 클래스

### 📋 과제
`Vehicle` 부모 클래스와 `Car`, `Bicycle` 자식 클래스를 만드세요.

**Vehicle**: `__init__(name, max_speed)`, `info()` 메소드  
**Car(Vehicle)**: `refuel()` — `"<이름>: <연료>를 주유합니다."`  
**Bicycle(Vehicle)**: `shift()` — `"<이름>: <gears>단 기어 변속합니다."`

(이번 Lab에서는 자식 클래스의 `__init__`을 정의하지 않습니다. `fuel`, `gears`는 인스턴스에 직접 할당)

### ✅ 예상 출력
```
소나타의 최고 속도는 200km/h입니다.
소나타: 휘발유를 주유합니다.
삼천리의 최고 속도는 40km/h입니다.
삼천리: 21단 기어 변속합니다.
```

````python
class Vehicle:
    def __init__(self, name, max_speed):
        # TODO 10-1
        self.name = name
        self.max_speed = max_speed
    
    def info(self):
        # TODO 10-2: f"{name}의 최고 속도는 {max_speed}km/h입니다."
        print(f"{self.name}의 최고 속도는 {self.max_speed}km/h입니다.")


class Car(Vehicle):
    # __init__ 없음 (부모 것 사용)
    def refuel(self):
        # TODO 10-3
        print(f"{self.name}: {self.fuel}를 주유합니다.")


class Bicycle(Vehicle):
    def shift(self):
        # TODO 10-4
        print(f"{self.name}: {self.gears}단 기어 변속합니다.")


# 테스트
my_car = Car("소나타", 200)
my_car.fuel = "휘발유"
my_car.info()
my_car.refuel()

my_bike = Bicycle("삼천리", 40)
my_bike.gears = 21
my_bike.info()
my_bike.shift()
````

## 🧪 Lab 11. `super()` 함수

### 📋 과제
Lab 10을 개선하여 각 자식 클래스가 자신의 `__init__`을 갖도록 하세요. `ElectricCar`도 추가합니다.

**Car(Vehicle)**: `__init__(name, max_speed, fuel)`  
**Bicycle(Vehicle)**: `__init__(name, max_speed, gears)`  
**ElectricCar(Vehicle)**: `__init__(name, max_speed, battery_kwh)`, `charge()` 메소드

### ✅ 예상 출력
```
아이오닉의 최고 속도는 180km/h입니다.
아이오닉: 배터리 77kWh를 충전합니다.
```

````python
class Vehicle:
    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed
    
    def info(self):
        print(f"{self.name}의 최고 속도는 {self.max_speed}km/h입니다.")


class Car(Vehicle):
    def __init__(self, name, max_speed, fuel):
        # TODO 11-1: super().__init__(...) 후 self.fuel 할당
        super().__init__(name, max_speed)
        self.fuel = fuel
    
    def refuel(self):
        print(f"{self.name}: {self.fuel}를 주유합니다.")


class Bicycle(Vehicle):
    def __init__(self, name, max_speed, gears):
        # TODO 11-2
        super().__init__(name, max_speed)
        self.gears = gears
    
    def shift(self):
        print(f"{self.name}: {self.gears}단 기어 변속합니다.")


class ElectricCar(Vehicle):
    def __init__(self, name, max_speed, battery_kwh):
        # TODO 11-3
        super().__init__(name, max_speed)
        self.battery_kwh = battery_kwh
    
    def charge(self):
        # TODO 11-4: "이름: 배터리 ○○kWh를 충전합니다."
        print(f"{self.name}: 배터리 {self.battery_kwh}kWh를 충전합니다.")


# 테스트
ev = ElectricCar("아이오닉", 180, 77)
ev.info()
ev.charge()
````

```
=== isinstance 테스트 ===
isinstance(my_car, Car): True
isinstance(my_car, Vehicle): True
isinstance(my_car, Bicycle): False
isinstance(ev, Vehicle): True

=== issubclass 테스트 ===
issubclass(ElectricCar, Vehicle): True
issubclass(Car, ElectricCar): False
issubclass(Car, Vehicle): True
```

⚠️ 먼저 Lab 11 셀을 실행해서 `Car`, `ElectricCar` 등이 메모리에 있어야 합니다.

## 🏁 도전 과제 (선택)

학습한 모든 개념을 활용한 미니 도서관 시스템입니다.

**요구사항**
1. `Item` 부모 클래스 (속성: `title`, `_available`, 클래스 변수 `library_name`)
2. `Book(Item)`: 추가 속성 `author`, `pages`
3. `DVD(Item)`: 추가 속성 `director`, `runtime`
4. 모든 클래스 `__str__` 정의
5. `borrow()`, `return_item()` 메소드 — `_available` 상태에 따라 분기
6. `isinstance()`로 책과 DVD를 분류하는 `categorize()` 함수

### 💡 힌트
- 부모 `__str__`을 자식에서 활용하려면 `super().__str__()` 사용
- `categorize(items)`는 `(books, dvds)` 튜플 반환

````python
class Item:
    # TODO C-1: 클래스 변수 library_name = "중앙도서관"
    library_name = "중앙도서관"
    
    def __init__(self, title):
        # TODO C-2: self.title, self._available = True
        self.title = title
        self._available = True
    
    def borrow(self):
        # TODO C-3: _available이 True면 False로 변경 + 대출완료 메시지
        #          False면 "이미 대출중" 메시지
        if self._available:
            self._available = False
            print(f"[{self.title}] 대출 완료")
        else:
            print(f"[{self.title}] 이미 대출중")
    
    def return_item(self):
        # TODO C-4: 위와 반대 동작
        if self._available:
            print(f"[{self.title}] 이미 반납 상태")
        else:
            self._available = True
            print(f"[{self.title}] 반납 완료")
    
    def __str__(self):
        # TODO C-5: "[제목] - 대출가능/대출중" 형식
        status = "대출가능" if self._available else "대출중"
        return f"[{self.title}] - {status}"


class Book(Item):
    def __init__(self, title, author, pages):
        # TODO C-6
        super().__init__(title)
        self.author = author
        self.pages = pages
    
    def __str__(self):
        # TODO C-7: 부모 __str__ + (저자: ..., ○○p)
        return f"{super().__str__()} (저자: {self.author}, {self.pages}p)"


class DVD(Item):
    def __init__(self, title, director, runtime):
        # TODO C-8
        super().__init__(title)
        self.director = director
        self.runtime = runtime
    
    def __str__(self):
        # TODO C-9
        return f"{super().__str__()} (감독: {self.director}, {self.runtime}분)"


def categorize(items):
    # TODO C-10: isinstance를 사용해 books, dvds로 분류
    books = []
    dvds = []
    for item in items:
        if isinstance(item, Book):
            books.append(item)
        elif isinstance(item, DVD):
            dvds.append(item)
    return books, dvds


# 테스트
items = [
    Book("파이썬 입문", "김교수", 350),
    DVD("기생충", "봉준호", 132),
    Book("객체 지향의 사실과 오해", "조영호", 224),
    DVD("올드보이", "박찬욱", 120),
]

for it in items:
    print(it)

print()
items[0].borrow()
items[0].borrow()
items[0].return_item()

print()
books, dvds = categorize(items)
print(f"책 {len(books)}권, DVD {len(dvds)}편")
````
